#¿Qué es un DataFrame de pandas?

Un DataFrame es una estructura de datos bidimensional (como una tabla) de la biblioteca pandas en Python. Piensa en él como el equivalente a una hoja de cálculo de Excel o a una tabla de SQL, pero dentro de Python.

Analogía simple: Imagina que un DataFrame es como una tabla con:

    * Filas: registros o observaciones individuales
    * Columnas: variables o características
    * Índices: identificadores únicos para cada fila


#Características más importantes

````python
import sqlite3
import pandas as pd

# Archivos públicos del repositorio de ejemplo
base = "https://raw.githubusercontent.com/iesataulfoargentasor/AlmacenDatosFakerIABD/main"

# Leer desde diferentes fuentes
df_csv = pd.read_csv(f"{base}/archivo.csv")
df_excel = pd.read_excel(f"{base}/archivo.xlsx")

# SQLite en memoria: no hace falta un servidor.
# El CSV se carga en una tabla y se consulta con SQL.
conexion = sqlite3.connect(":memory:")
df_csv.to_sql("tabla", conexion, if_exists="replace", index=False)
df_sql = pd.read_sql("SELECT * FROM tabla", conexion)

print("CSV")
print(df_csv.head())
print("\nExcel")
print(df_excel.head())
print("\nSQL")
print(df_sql.head())

# Guardar datos (df viene de las celdas anteriores)
df.to_csv("nuevo_archivo.csv", index=False)
df.to_excel("nuevo_archivo.xlsx", index=False)

conexion.close()
````

#Ejemplo práctico completo
```python

import pandas as pd

# Crear DataFrame
ventas = pd.DataFrame({
    'Producto': ['A', 'B', 'A', 'C', 'B', 'C'],
    'Cantidad': [10, 15, 8, 12, 20, 5],
    'Precio': [100, 150, 100, 200, 150, 200]
})

# Calcular venta total por producto
ventas['Total'] = ventas['Cantidad'] * ventas['Precio']
ventas_por_producto = ventas.groupby('Producto')['Total'].sum()

print(ventas_por_producto)
```
Los DataFrames de pandas son fundamentales para el análisis de datos en Python y son la base para muchas operaciones en ciencia de datos y machine learning.

````python
import pandas as pd

# Crear DataFrame
ventas = pd.DataFrame({
    'Producto': ['A', 'B', 'A', 'C', 'B', 'C'],
    'Cantidad': [10, 15, 8, 12, 20, 5],
    'Precio': [100, 150, 100, 200, 150, 200]
})

# Calcular venta total por producto
ventas['Total'] = ventas['Cantidad'] * ventas['Precio']
ventas_por_producto = ventas.groupby('Producto')['Total'].sum()

print(ventas_por_producto)
````
